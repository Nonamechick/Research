```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# 1. Define a system of differential equations (Lorenz Attractor)
def lorenz(t, state, sigma=10.0, beta=8/3, rho=28.0):
    x, y, z = state
    return [sigma * (y - x), x * (rho - z) - y, x * y - beta * z]

# 2. Numerical Integration using SciPy
t_span = (0, 40)
t_eval = np.linspace(t_span[0], t_span[1], 10000)
initial_state = [1.0, 1.0, 1.0]

solution = solve_ivp(lorenz, t_span, initial_state, t_eval=t_eval)
x, y, z = solution.y

# 3. Plotting results using Matplotlib
fig = plt.figure(figsize=(10, 6), dpi=100)
ax = fig.add_subplot(111, projection='3d')

# Plot chaotic trajectory
ax.plot(x, y, z, lw=0.6, color='crimson')
ax.set_title("Lorenz Attractor - System Verification Test", fontsize=14, fontweight='bold')
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.grid(True)

plt.show()
```
